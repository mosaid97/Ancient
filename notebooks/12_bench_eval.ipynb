# 12 — Bench eval (AncientChinaSearch-Bench)

Thin wrapper around `eval.harness.run` — runs the full bench (NDCG@10, Recall@k, faithfulness, linking) with 10k-bootstrap CIs and persists the artifact for the README badge.

```python
import json
import sys
from datetime import datetime, timezone
from pathlib import Path

REPO_ROOT = Path.cwd()
while REPO_ROOT != REPO_ROOT.parent and not (REPO_ROOT / 'apps').exists():
    REPO_ROOT = REPO_ROOT.parent
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))

from dotenv import load_dotenv
load_dotenv(REPO_ROOT / '.env')
from eval.harness import run as run_harness

metrics = run_harness(REPO_ROOT / 'data' / 'bench', REPO_ROOT / 'eval_out', k=10)
ART = REPO_ROOT / 'notebooks' / '_artifacts' / '12_bench_eval'
ART.mkdir(parents=True, exist_ok=True)
(ART / 'metrics.json').write_text(json.dumps(metrics, ensure_ascii=False, indent=2, default=str))
print('artifact:', ART / 'metrics.json')
```
